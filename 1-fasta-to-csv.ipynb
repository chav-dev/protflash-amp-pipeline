# 🧬 1. Fasta to CSV Converter

Este notebook forma parte de la etapa inicial de preparación de datos del proyecto **ProtFlash**. Su propósito principal es transformar archivos de secuencias biológicas de un formato estándar a un formato tabular.

---

## 📋 Descripción General
El script realiza la **conversión de archivos FASTA** (secuencias de proteínas) a **formato CSV**. Esta transformación es fundamental para facilitar el análisis exploratorio y el preprocesamiento necesario antes de alimentar modelos de Machine Learning utilizando librerías como `pandas`.



## 🛠️ Funcionalidades Principales

1.  **Configuración de Entorno:**
    * Define rutas personalizadas en Google Drive para la entrada (`/fasta`) y salida (`/csv`).
2.  **Instalación de Dependencias:**
    * Configura el entorno de Colab instalando `biopython`, la librería estándar para el manejo de datos bioinformáticos.
3.  **Lógica de Extracción:**
    * La función `fasta_to_csv` extrae de cada entrada:
        * `ID`: Identificador de la proteína.
        * `Description`: Metadatos del encabezado.
        * `Sequence`: La cadena de aminoácidos.
        * `Length`: Longitud calculada de la secuencia.
4.  **Procesamiento por Lotes:**
    * Utiliza `glob` para identificar y procesar automáticamente todos los archivos `.fasta` presentes en el directorio de origen, manteniendo la organización de los datos.

## 📁 Estructura de Salida
Los archivos resultantes se guardan con el mismo nombre original pero con extensión `.csv`, listos para la siguiente etapa de **Feature Extraction**.

Configuración de rutas

```python
INPUT_FASTA = "/content/drive/MyDrive/Experimentos/ProtFlash/Data/fasta"
OUTPUT_CSV = "/content/drive/MyDrive/Experimentos/ProtFlash/Data/csv"
```

Importacion de librerias

```python
from Bio import SeqIO
import pandas as pd
import os
import glob
```

Conversión FASTA → CSV

```python
def fasta_to_csv(input_fasta, output_csv):
    """Convierte un archivo FASTA a CSV con columnas: id, description, sequence, length"""
    records = []
    for record in SeqIO.parse(input_fasta, "fasta"):
        records.append({
            "id": record.id,
            "description": record.description,
            "sequence": str(record.seq),
            "length": len(record.seq)
        })

    df = pd.DataFrame(records)
    df.to_csv(output_csv, index=False)
    print(f"✅ {os.path.basename(input_fasta)} → {os.path.basename(output_csv)} ({len(df)} secuencias)")
    return df
```

```python
fasta_files = glob.glob(os.path.join(INPUT_FASTA, "*.fasta"))
```

```python
if not fasta_files:
    print(f"⚠️ No se encontraron archivos .fasta en: {INPUT_FASTA}")
else:
    for fasta_path in sorted(fasta_files):
        nombre = os.path.splitext(os.path.basename(fasta_path))[0]
        csv_path = os.path.join(OUTPUT_CSV, f"{nombre}.csv")
        fasta_to_csv(fasta_path, csv_path)
```
